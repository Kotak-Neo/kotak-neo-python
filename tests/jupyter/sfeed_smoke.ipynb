```python
import asyncio

print("event loop running:", asyncio.get_running_loop())


async def try_connect():
    ws2 = SFeedWebSocket(
        access_token="dummy",
        sid="1",
        ucc="TESTUCC",
        url="wss://invalid.invalid/wsfeed",
        max_connect_retries=0,
    )
    try:
        await ws2.connect()
    except Exception as e:
        return f"caught: {type(e).__name__}: {e}"
    return "connected (unexpected)"


result = await try_connect()
print(result)
```
